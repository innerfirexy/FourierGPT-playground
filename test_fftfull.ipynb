```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
human_real = pd.read_csv('data/pubmed_gpt-4.original.gpt2.nllzs.fftreal.txt')
human_imag = pd.read_csv('data/pubmed_gpt-4.original.gpt2.nllzs.fftimag.txt')

human_full = pd.concat([human_real, human_imag['power']], axis=1)
human_full.columns = ['sid', 'freq', 'real', 'imag']
human_full
```

```python
model_real = pd.read_csv('data/pubmed_gpt-4.sampled.gpt2.nllzs.fftreal.txt')
model_imag = pd.read_csv('data/pubmed_gpt-4.sampled.gpt2.nllzs.fftimag.txt')

model_full = pd.concat([model_real, model_imag['power']], axis=1)
model_full.columns = ['sid', 'freq', 'real', 'imag']
model_full
```

```python
human_subset = human_full
model_subset = model_full

fig, ax = plt.subplots()
ax.scatter(human_subset['real'], human_subset['imag'], marker='o', color='red')
ax.scatter(model_subset['real'], model_subset['imag'], marker='x', color='blue')
plt.show()
```

```python
print(np.mean(human_full['real']), np.std(human_full['real']), np.mean(human_full['imag']), np.std(human_full['imag']))
print(np.mean(model_full['real']), np.std(model_full['real']), np.mean(model_full['imag']), np.std(model_full['imag']))
```

```python
human_full['power'] = np.sqrt(human_full['real']**2 + human_full['imag']**2)
model_full['power'] = np.sqrt(model_full['real']**2 + model_full['imag']**2)
```

```python
fig, ax = plt.subplots()
ax.scatter(human_full['freq'], human_full['power'], marker='o', color='red')
ax.scatter(model_full['freq'], model_full['power'], marker='x', color='blue')
plt.show()
```

```python
print(np.mean(human_full['power']), np.std(human_full['power']))
print(np.mean(model_full['power']), np.std(model_full['power']))
```
